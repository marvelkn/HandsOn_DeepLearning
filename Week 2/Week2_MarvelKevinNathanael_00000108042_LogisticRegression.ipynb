Satu Dense layer sebagai baseline regresi logistik multikelas sesuai TLM 2.

```python
import numpy as np
import tensorflow as tf
from sklearn.model_selection import train_test_split
from sklearn.metrics import confusion_matrix, classification_report

tf.keras.utils.set_random_seed(42)
(x_dev, y_dev), (x_test, y_test) = tf.keras.datasets.mnist.load_data()
x_dev = x_dev.astype("float32") / 255.0
x_test = x_test.astype("float32") / 255.0
x_train, x_val, y_train, y_val = train_test_split(
    x_dev, y_dev, test_size=0.1, random_state=42, stratify=y_dev)
model = tf.keras.Sequential([
    tf.keras.Input(shape=(28, 28)), tf.keras.layers.Flatten(),
    tf.keras.layers.Dense(10, activation="softmax")
])
model.compile(optimizer="adam", loss="sparse_categorical_crossentropy",
              metrics=["accuracy"])
history = model.fit(x_train, y_train, validation_data=(x_val, y_val),
                    epochs=15, batch_size=128, verbose=1,
                    callbacks=[tf.keras.callbacks.EarlyStopping(
                        monitor="val_loss", patience=3, restore_best_weights=True)])
```

```python
loss, accuracy = model.evaluate(x_test, y_test, verbose=0)
y_pred = np.argmax(model.predict(x_test, verbose=0), axis=1)
print(f"Test accuracy: {accuracy:.4f}")
print(classification_report(y_test, y_pred, digits=3))
print("Confusion matrix:\n", confusion_matrix(y_test, y_pred))
```

**Catatan:** Output softmax sudah berupa probabilitas, sehingga loss tidak memakai from_logits=True. Periksa confusion matrix untuk kesalahan per digit.
