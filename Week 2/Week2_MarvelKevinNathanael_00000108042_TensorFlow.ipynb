```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split

rng = np.random.default_rng(42)
area = 2.5 * rng.normal(size=100) + 25
price = 25 * area + 5 + rng.integers(20, 50, size=100)
x_train, x_test, y_train, y_test = train_test_split(
    area, price, test_size=0.2, random_state=42)
print("Train:", len(x_train), "| Test:", len(x_test))

import tensorflow as tf
tf.keras.utils.set_random_seed(42)
```

```python
normalizer = tf.keras.layers.Normalization(axis=-1)
normalizer.adapt(x_train.reshape(-1, 1))  # Statistik hanya dari train.
model = tf.keras.Sequential([tf.keras.Input(shape=(1,)), normalizer,
                             tf.keras.layers.Dense(1)])
model.compile(optimizer=tf.keras.optimizers.Adam(learning_rate=0.05),
              loss="mse", metrics=["mae"])
history = model.fit(x_train.reshape(-1, 1), y_train, validation_split=0.2,
                    epochs=200, batch_size=16, verbose=0,
                    callbacks=[tf.keras.callbacks.EarlyStopping(
                        monitor="val_loss", patience=20, restore_best_weights=True)])
test_mse, test_mae = model.evaluate(x_test.reshape(-1, 1), y_test, verbose=0)
print(f"Epoch: {len(history.history['loss'])} | Test MAE: {test_mae:.3f} | Test RMSE: {np.sqrt(test_mse):.3f}")
pred = model.predict(x_test.reshape(-1, 1), verbose=0).ravel()
order = np.argsort(x_test)
plt.scatter(x_test, y_test, label="Test")
plt.plot(x_test[order], pred[order], color="red", label="Prediksi")
plt.xlabel("Area"); plt.ylabel("Price"); plt.legend(); plt.show()
```

**Catatan:** Satu neuron mempelajari fungsi linear yang sama seperti OLS. Optimisasi gradien dapat memberi hasil sedikit berbeda.
