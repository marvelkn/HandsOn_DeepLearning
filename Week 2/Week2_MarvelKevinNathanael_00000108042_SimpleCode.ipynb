Contoh harga terhadap luas rumah dari TLM 2. Data sintetis hanya untuk memeriksa perhitungan koefisien.

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split

rng = np.random.default_rng(42)
area = 2.5 * rng.normal(size=100) + 25
price = 25 * area + 5 + rng.integers(20, 50, size=100)
x_train, x_test, y_train, y_test = train_test_split(
    area, price, test_size=0.2, random_state=42)
print("Train:", len(x_train), "| Test:", len(x_test))
```

```python
w = np.sum((x_train - x_train.mean()) * (y_train - y_train.mean())) / np.sum((x_train - x_train.mean()) ** 2)
b = y_train.mean() - w * x_train.mean()
y_pred = w * x_test + b
print(f"price = {w:.3f} * area + {b:.3f}")
print(f"Test MAE: {np.mean(np.abs(y_test - y_pred)):.3f}")
print(f"Test RMSE: {np.sqrt(np.mean((y_test - y_pred) ** 2)):.3f}")
order = np.argsort(x_test)
plt.scatter(x_train, y_train, s=18, label="Train")
plt.scatter(x_test, y_test, s=22, label="Test")
plt.plot(x_test[order], y_pred[order], color="red", label="Prediksi")
plt.xlabel("Area"); plt.ylabel("Price"); plt.legend(); plt.show()
```

**Catatan:** Regresi biasa cukup untuk pola linear ini. Error test lebih informatif daripada garis pada train.
